# Jumeau numérique — Random Forest


## 1. Import des bibliothèques

```python
import argparse
import json
from pathlib import Path

import matplotlib
%matplotlib inline
import matplotlib.pyplot as plt
from IPython.display import display
import numpy as np
import pandas as pd
import seaborn as sns
import joblib
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix, f1_score, log_loss)
from sklearn.model_selection import train_test_split
```

## 2. Collecte et affichage des données

```python
dataset = pd.read_csv("digitalTwin.csv")
```

## 3. Nettoyage des données

```python
print("Doublons :", dataset.duplicated().sum())
print(dataset.isnull().sum())
```

```python
dataset = dataset.drop_duplicates()
dataset = dataset.dropna()
colonnes_numeriques = ["HealthIndex", "EngineTemp", "Vibration", "Pressure"]
dataset[colonnes_numeriques] = dataset[colonnes_numeriques].apply(pd.to_numeric, errors="coerce")
dataset = dataset.dropna(subset=colonnes_numeriques)
dataset = dataset[(dataset[colonnes_numeriques] >= 0).all(axis=1)]
```

```python
print("Dimensions après nettoyage :", dataset.shape)
print(dataset.isnull().sum())
```

## 4. Analyse exploratoire (EDA)

```python
print(dataset["HealthIndex"].describe().round(2))
print(dataset.groupby("State")[colonnes_numeriques].mean().round(2))
```

```python
sns.set_theme(style="whitegrid")
figure, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=dataset, x="State", order=dataset["State"].value_counts().index, ax=axes[0])
axes[0].set_title("Répartition des états")
sns.boxplot(data=dataset, x="State", y="HealthIndex", ax=axes[1])
axes[1].set_title("HealthIndex selon l'état")
figure.tight_layout()
display(figure)
plt.close(figure)
```

```python
correlations = dataset[colonnes_numeriques].corr()
figure, axe = plt.subplots(figsize=(8, 6))
sns.heatmap(correlations, annot=True, cmap="coolwarm", vmin=-1, vmax=1, ax=axe)
axe.set_title("Matrice de corrélation")
display(figure)
plt.close(figure)
```

## 5. Ingénierie des caractéristiques

```python
dataset["TempVibration"] = dataset["EngineTemp"] * dataset["Vibration"]
dataset["PressureVibrationRatio"] = dataset["Pressure"] / (dataset["Vibration"] + 0.01)
dataset["HealthRisk"] = 100 - dataset["HealthIndex"]
```

```python
features = ["HealthIndex", "EngineTemp", "Vibration", "Pressure", "TempVibration", "PressureVibrationRatio", "HealthRisk"]
X = dataset[features]
y = dataset["State"]
```

```python
print(X.head())
print(y.value_counts())
```

## 6. Entraînement du modèle Random Forest

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
modele_random_forest = RandomForestClassifier(n_estimators=300, max_depth=12, min_samples_leaf=2, random_state=42, n_jobs=-1, class_weight="balanced")
```

```python
modele_random_forest.fit(X_train, y_train)
```

```python
joblib.dump(modele_random_forest, "random_forest_state_model.joblib")
```

## 7. Évaluation du modèle

```python
predictions = modele_random_forest.predict(X_test)
probabilites = modele_random_forest.predict_proba(X_test)
```

```python
print("Accuracy Random Forest :", round(accuracy_score(y_test, predictions), 4))
print("F1 pondéré Random Forest :", round(f1_score(y_test, predictions, average="weighted"), 4))
print("Log loss Random Forest :", round(log_loss(y_test, probabilites, labels=modele_random_forest.classes_), 4))
print(classification_report(y_test, predictions))
```

```python
matrice = confusion_matrix(y_test, predictions, labels=modele_random_forest.classes_)
figure, axe = plt.subplots(figsize=(7, 5))
sns.heatmap(matrice, annot=True, fmt="d", cmap="Greens", xticklabels=modele_random_forest.classes_, yticklabels=modele_random_forest.classes_, ax=axe)
axe.set_xlabel("Prédiction"); axe.set_ylabel("Valeur réelle"); axe.set_title("Matrice de confusion - Random Forest")
display(figure)
plt.close(figure)
```

```python
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
print("Accuracy baseline :", round(accuracy_score(y_test, baseline.predict(X_test)), 4))
```

## 8. Explicabilité : importance Random Forest, SHAP et LIME

```python
importances = pd.Series(modele_random_forest.feature_importances_, index=features).sort_values()
print(importances.sort_values(ascending=False))
axe = importances.plot(kind="barh", color="forestgreen", title="Importance des variables - Random Forest")
figure = axe.get_figure()
figure.tight_layout()
display(figure)
plt.close(figure)

# Si nécessaire, exécuter une fois : %pip install shap lime
```
